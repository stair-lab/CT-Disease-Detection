```python
# Create a copy of the original DataFrame to process
df_copy = df.copy()

# Identify columns with mixed types (object or mixed-type columns)
mixed_cols = df_copy.columns[df_copy.apply(lambda col: col.apply(lambda x: isinstance(x, str) or pd.isnull(x)).any())]

# Track columns that successfully convert to numeric
converted_cols = []

# Convert mixed-type columns to numeric where possible
for col in mixed_cols:
    try:
        # Attempt conversion and create a temporary column to check success
        temp_col = pd.to_numeric(df_copy[col], errors="coerce")
        
        # Check if at least 50% of the rows were successfully converted
        if temp_col.notnull().mean() > 0.5:  # Keep only if >50% rows are valid
            df_copy[col] = temp_col
            converted_cols.append(col)
        else:
            print(f"Skipping conversion for column '{col}' due to too many NaNs.")
    except Exception as e:
        print(f"Error converting column '{col}': {e}")

# Drop rows only where converted columns contain NaN (to prevent massive row loss)
df_copy = df_copy.dropna(subset=converted_cols)

# Select only numerical columns from the copy for plotting
numeric_cols = df_copy.select_dtypes(include=["number"]).columns

# Sort the numeric columns in alphabetical order
numeric_cols = sorted(numeric_cols)

# Number of numeric columns
num_cols = len(numeric_cols)

# Set plot style
sns.set(style="whitegrid")

# Determine the number of rows and columns for the panel
num_rows = (num_cols + 2) // 3  # Create rows of 3 plots per row

# Create a figure and axis grid
fig, axes = plt.subplots(num_rows, 3, figsize=(15, 5 * num_rows))
axes = axes.flatten()

# Plot frequency distributions for numerical columns
for i, col in enumerate(numeric_cols):
    sns.histplot(df_copy[col], kde=True, ax=axes[i], color="skyblue", edgecolor="black")
    axes[i].set_title(f"{col} Distribution")
    axes[i].set_xlabel("")
    axes[i].set_ylabel("Frequency")

# Hide any unused subplots
for j in range(i + 1, len(axes)):
    fig.delaxes(axes[j])

# Adjust layout for better spacing
plt.tight_layout()
plt.show()
```

```python
# Generate a report for the number of NaN values in each column
null_report = df.isnull().sum()

# Sort the columns alphabetically
sorted_columns = sorted(null_report.index)

# Print the result for each column in alphabetical order
print(f"Number of Total Rows in the DataFrame:{len(df)}")
print("Number of NaN values in each column (sorted alphabetically):")
for column in sorted_columns:
    print(f"{column}: {null_report[column]}")
```
